```python
# 1. Множества V,E
g_v = [1, 2, 3, 4]
g_e = {'e1':(1,3), 'e2':(1,2), 'e3':(3,4), 'e4':(4,2), 'e5':(3,2), 'e6':(2,2), 'e7':(1,2)}
print(f'1. Множества V,E: \n{g_v}\n{g_e}')

# 2. Пары смежных вершин
def adj_v_check(v,e):
  adj_ = []
  seen = set()
  for k, (u, w) in e.items():
      if u != w:
          p = tuple(sorted((u, w)))
          if p not in seen:
              adj_.append(p)
              seen.add(p)
  return adj_
adj_v = adj_v_check(g_v,g_e)
print(f'2. Пары смежных вершин: {adj_v}')

# 3. Инцидентность ребра вершинам
print(f'3. Инцидентность ребра вершинам: {g_e}')

# 4. Пары смежных рёбер
def adj_e_check(v,e):
  adj_ = []
  ek = list(e.keys())
  for i in range(len(ek)):
      for j in range(i + 1, len(ek)):
          e1, e2 = e[ek[i]], e[ek[j]]
          if set(e1) & set(e2): adj_.append((ek[i], ek[j]))
  return adj_
adj_e = adj_e_check(g_v,g_e)
print(f'4. Пары смежных рёбер: {adj_e}')

# 5. Степени вершин
def deg_check(v,e):
  deg = {n: 0 for n in v}
  for u, w in e.values():
      if u == w: deg[u] += 2
      else:
          deg[u] += 1
          deg[w] += 1
  return deg
deg = deg_check(g_v,g_e)
print(f'5. Степени вершин: {deg}')

# 6. Параллельные рёбра
def par_check(v,e):
  par = []
  seen_p = {}
  for k, edge in e.items():
      p = tuple(sorted(edge))
      if edge[0] == edge[1]: continue
      seen_p.setdefault(p, []).append(k)
  return [v for v in seen_p.values() if len(v) > 1]
par = par_check(g_v,g_e)
print(f'6. Параллельные рёбра: {par}')

# 7. Наличие петель
def loop_check(v,e):
  return [k for k, edge in e.items() if edge[0] == edge[1]]
loops = loop_check(g_v,g_e)
print(f'7. Наличие петель: {f"да, {loops}" if loops else "нет"}')
```

```python
# Матрицы смежности и инцидентности:
def matrice_create_inc_adj(v,e):
  n = len(v)
  vm = {val: i for i, val in enumerate(v)}
  am = np.zeros((n, n), dtype=int)
  im = np.zeros((n, len(e)), dtype=int)
  for j, (k, (u, w)) in enumerate(e.items()):
      iu, iw = vm[u], vm[w]
      if u == w:
          am[iu][iu] += 1
          im[iu][j] = 2
      else:
          am[iu][iw] += 1
          am[iw][iu] += 1
          im[iu][j] = 1
          im[iw][j] = 1
  return am, im
g_v = [1, 2, 3, 4, 5, 6, 7]
g_e = {'e1':(1,1), 'e2':(1,2), 'e3':(2,3), 'e4':(1,3), 'e5':(4,6), 'e6':(4,5), 'e7':(5,6)}
am,im = matrice_create_inc_adj(g_v,g_e)
print(f'Матица смежности: \n{am} \nМатрица инцидентности:\n{im}')

# Из задания 1
print(f'\nГраф задания 1:')
g_v = [1, 2, 3, 4]
g_e = {'e1':(1,3), 'e2':(1,2), 'e3':(3,4), 'e4':(4,2), 'e5':(3,2), 'e6':(2,2), 'e7':(1,2)}
am,im = matrice_create_inc_adj(g_v,g_e)
print(f'Матица смежности: \n{am} \nМатрица инцидентности:\n{im}')
```

```python
# 1. Множества V,E
g_v = [1, 2, 3, 4, 5]
g_e = {'e1':(4,1), 'e2':(1,4), 'e3':(1,2), 'e4':(3,2), 'e5':(2,3), 'e6':(3,3), 'e7':(3,4)}
print(f'1. Множества V,E: \n{g_v}\n{g_e}')

# 2. Пары смежных вершин
def adj_v_check(v,e):
  return list({tuple(sorted(edge)) for edge in e.values() if edge[0] != edge[1]})
adj_v = adj_v_check(g_v,g_e)
print(f'2. Пары смежных вершин: {adj_v}')

# 3. Наличие петель
def loop_check(v,e):
  return [k for k, edge in e.items() if edge[0] == edge[1]]
loops = loop_check(g_v,g_e)
print(f'3. Наличие петель: {f"да, {loops}" if loops else "нет"}')

# 4. Степени вершин
def deg_check(v,e):
  deg = {n: 0 for n in v}
  for u, w in e.values():
      if u == w: deg[u] += 2
      else: deg[u] += 1; deg[w] += 1
  return deg
deg = deg_check(g_v,g_e)
print(f'4. Степени вершин: {deg}')

# 5. Количество вершин нечетной степени
def odd_check(v,e,deg):
  return sum(1 for d in deg.values() if d % 2 != 0)
odds = odd_check(g_v,g_e,deg)
print(f'5. Количество вершин нечетной степени: {odds}')

def matrice_create_inc_adj(v,e):
  n, m = len(v), len(e)
  vm = {val: i for i, val in enumerate(v)}
  am = np.zeros((n, n), dtype=int)
  im = np.zeros((n, m), dtype=int)
  for j, (u, w) in enumerate(e.values()):
      iu, iw = vm[u], vm[w]
      am[iu][iw] += 1
      if u == w: im[iu][j] = 0 # loops in directed incidence often 0 or special
      else:
          im[iu][j] = -1
          im[iw][j] = 1
  return am, im
am,im = matrice_create_inc_adj(g_v,g_e)
print(f'6. Матрица инцидентности:\n{im}\n7. Матрица смежности: \n{am} \n')
```

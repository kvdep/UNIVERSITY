###классификация

```python
p = load_breast_cancer(as_frame=True).frame
display(p.head())
display(p.info())
display(p.describe())
```

```python
p = load_breast_cancer()
x = p.data
y = p.target

def analyse (x,y,m):
  x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8)
  m.fit(x_train,y_train)
  #y_train_pred = m.predict(x_train)
  y_test_pred = m.predict(x_test)
  #print(cm(y_train, y_train_pred))
  print(cm(y_test, y_test_pred))
  metrics = pd.DataFrame({
    "Test": [
        acc(y_test, y_test_pred),
        pr(y_test, y_test_pred,average='macro'),
        rec(y_test, y_test_pred,average='macro'),
        f1(y_test, y_test_pred,average='macro'),
    ],
    }, index = ["Accuracy", "Precision", "Recall", "F1"])
  display(metrics)
#1
print(f'KNN:')
m = KNN(n_neighbors=7)
analyse(x,y,m)
print(f'____________\n\n')
#2
print('LogReg:')
m = LRC(max_iter=10000)
analyse(x,y,m)
print(f'____________\n\n')
#3
print('Tree:')
m = Tree(max_depth = 10)
analyse(x,y,m)
print(f'____________\n\n')
#4
print('SVM:')
m = SVC(kernel = 'linear')
analyse(x,y,m)
print(f'____________\n\n')
#5
print('Random Forest:')
m = RFC()
analyse(x,y,m)
print(f'____________\n\n')
```

##2 Повторите анализ для другого датасета по вашему выбору. Используйте несколько моделей для сравнения. Используйте датасет для множественной классификации.

```python
p = load_wine(as_frame=True).frame
display(p.head())
display(p.info())
display(p.describe())
```

```python
p = load_wine(as_frame=True).frame
x = p.drop('target',axis=1)
y = p[['target']]

def analyse (x,y,m):
  x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8)
  m.fit(x_train,y_train)
  y_test_pred = m.predict(x_test)
  print(cm(y_test, y_test_pred))
  metrics = pd.DataFrame({
    "Test": [
        acc(y_test, y_test_pred),
        pr(y_test, y_test_pred,average='micro'),
        rec(y_test, y_test_pred,average='macro'),
        f1(y_test, y_test_pred,average='weighted'),
    ],
    }, index = ["Accuracy", "Precision", "Recall", "F1"])
  display(metrics)
#1
print(f'KNN:')
m = KNN(n_neighbors=7)
analyse(x,y,m)
print(f'____________\n\n')
#2
print('LogReg:')
m = LRC(max_iter=10000)
analyse(x,y,m)
print(f'____________\n\n')
#3
print('Tree:')
m = Tree(max_depth = 4)
analyse(x,y,m)
print(f'____________\n\n')
#4
print('SVM:')
m = SVC(kernel = 'linear')
analyse(x,y,m)
print(f'____________\n\n')
#5
print('Random Forest:')
m = RFC()
analyse(x,y,m)
print(f'____________\n\n')
```

```python
p = fetch_california_housing(as_frame=True)
x = p.data
y = p.target

def analyse (x,y,m):
  x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8)
  m.fit(x_train,y_train)
  y_test_pred = m.predict(x_test)
  metrics = pd.DataFrame({
    "Test": [
        r2(y_test, y_test_pred),
        mae(y_test, y_test_pred),
        mse(y_test, y_test_pred),
        mape(y_test, y_test_pred),
        rmse(y_test, y_test_pred),
        medae(y_test, y_test_pred),
        maxe(y_test, y_test_pred)
    ],
    }, index = ["R2", "MAE", "MSE","MAPE","RMSE",'MedianError','MaxError'])
  display(metrics)
  print(f'\n\n')



sc = SC()
x = sc.fit_transform(x)
#mms = MMS()
#x = mms.fit_transform(x)

m = LR()
print(f'Linear Regression')
analyse(x,y,m)

pf = PF(degree=2)
x1 = pf.fit_transform(x)
m = LR()
print(f'Polynomial Regression')
analyse(x1,y,m)

m = SVR(kernel='rbf',C=5,epsilon = 0.05)
print(f'SVR')
analyse(x,y,m)

m = Lasso()
print(f'Lasso (L1)')
analyse(x,y,m)

m = Ridge()
print(f'Ridge (L2)')
analyse(x,y,m)

m = ElasticNet(l1_ratio=0.3)
print(f'ElasticNet (L1 + L2)')
analyse(x,y,m)
```

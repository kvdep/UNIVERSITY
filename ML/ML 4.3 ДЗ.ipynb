#1. Изучите разбиение Leave-P-Out. Продемонстрируйте работу этого алгоритма на примере из лабораторной работы.

```python
x,y = load_iris(as_frame=True).data,load_iris().target
lpo = LeavePOut(2)
print(f'Столько сплитов: {lpo.get_n_splits(x)}')
for i, (train_index, test_index) in enumerate(lpo.split(x)):
    print(f"Fold {i}:")
    #print(f"  Train: index={train_index}")
    #print(f"  Test:  index={test_index}")
    m = LR().fit(x.values[train_index],y[train_index])
    print('Accuracy:',m.score(x.values[test_index],y[test_index]))
```

```python
p1 = pd.DataFrame(cross_validate(LR(),x.values,y,cv = 10,scoring=('r2', 'f1_macro','recall_macro')))
p1.columns = [f'{i}_LR' for i in p1.columns]
p2 = pd.DataFrame(cross_validate(RFC(),x.values,y,cv = 10,scoring=('r2', 'f1_macro','recall_macro')))
p2.columns = [f'{i}_RFC' for i in p2.columns]
p3 = p2.join(p1)
p3.head()
p3['fit_time_diff'] = p3['fit_time_RFC'] - p3['fit_time_LR']
p3['test_r2_diff'] = p3['test_r2_RFC'] - p3['test_r2_LR']
p3['test_f1_diff'] = p3['test_f1_macro_RFC'] - p3['test_f1_macro_LR']
print(f'RFC - LR:')
p3[['fit_time_diff','test_r2_diff' ,'test_f1_diff']]
```

#6. Сделайте k-блочную перекрёстную проверку (KFold) модели логистической регрессии, предварительно стандартизировав данные. Для этого создайте конвейер с помощью make_pipeline из библиотеки sklearn.pipeline, который будет стандартизировать, а затем выполнять логистическую регрессию.

```python
pipe = make_pipeline(SC(),LR())

folding = KFold(10, random_state=42,shuffle=True)

x,y = load_iris().data,load_iris().target

for i, (train_index, test_index) in enumerate(folding.split(x)):
    print(f"Fold {i}:")
    m = pipe.fit(x[train_index],y[train_index])
    print(f'Precision: {precision_score(y[test_index],m.predict(x[test_index]),average="macro")}')
    print(f'Recall:    {recall_score(y[test_index],m.predict(x[test_index]),average="macro")}')
    print()
```

```python
from sklearn.neural_network import MLPClassifier as MLPc
from sklearn.neural_network import MLPRegressor  as MLPr
from sklearn.svm import SVC
from sklearn.datasets import make_blobs as mb
import numpy as np
import pandas as pd
import time
```

###1 Создайте однослойный перцептрон с 1, 2, 10 и 100 нейронами. Сравние их точность и сделайте вывод о достаточном количестве нейронов.

```python
l = [1, 2, 10 , 100]
out = []

blob_centers = ([1, 1], [3, 4], [1, 3.3], [3.5, 1.8])
x, y = mb (n_samples=200,
                          centers=blob_centers,
                          cluster_std=0.5,
                          random_state=0)

for i in l:
  m = MLPc((i),max_iter=10000).fit(x,y)
  out.append((i,m.score(x,y)))
print(*out,sep='\n')
```

###2. Создайте и оцените модель с двумя, тремя и десятью скрытыми слоями с одинаковым количеством нейронов. Сравните их точность и сделайте вывод о достаточном количестве слоев.

```python
l = [2,3,10]
out = []

blob_centers = ([1, 1], [3, 4], [1, 3.3], [3.5, 1.8])
x, y = mb (n_samples=200,
                          centers=blob_centers,
                          cluster_std=0.5,
                          random_state=0)

for i in l:
  m = MLPc([3]*i,max_iter=10000).fit(x,y)
  out.append((i,m.score(x,y)))
print(*out,sep='\n')
```

###3. Для глубокой модели выведите веса всех нейронов на всех слоях. Выведите значения векторов весов смещения.

```python
blob_centers = ([1, 1], [3, 4], [1, 3.3], [3.5, 1.8])
x, y = mb (n_samples=200,
                          centers=blob_centers,
                          cluster_std=0.5,
                          random_state=0)

m = MLPc(hidden_layer_sizes=(10,10,10,10,10,10),max_iter=10000).fit(x,y)
m.score(x,y)
print('\n\nВеса:\n\n')
print(m.coefs_)

print(f'\n\nСмещения:\n\n')
print(m.intercepts_)
```

###4. Постройте и оцените модель с большим количеством нейронов и слоев. Замерьте время выполнения обучения, сравните со временем обучения более простых моделей.

```python
def timer(time1):
  print(f'{time.time()-time1} seconds')
```

```python
blob_centers = ([1, 1], [3, 4], [1, 3.3], [3.5, 1.8])
x, y = mb (n_samples=1000,
                          centers=blob_centers,
                          cluster_std=0.9,
                          random_state=0)

m = SVC()
time1=time.time()
m.fit(x,y)
print(f'SVC:')
timer(time1)
print(m.score(x,y))

time1=time.time()
m = MLPc(hidden_layer_sizes=(10,10,10,10,10,10),max_iter=10000).fit(x,y)
print(f'MLP, 6 слоев по 10 нейронов:')
timer(time1)
print(m.score(x,y))

time1=time.time()
m = MLPc(hidden_layer_sizes=(100,100,100,100,100,100),max_iter=10000).fit(x,y)
print(f'MLP, 6 слоев по 100 нейронов:')
timer(time1)
print(m.score(x,y))

time1=time.time()
m = MLPc(hidden_layer_sizes=(100,100,100,100,100,100,100,100,100,100,100,100,100,100,100,100,100,100),max_iter=10000).fit(x,y)
print(f'MLP, 20 слоев по 100 нейронов:')
timer(time1)
print(m.score(x,y))
```

###5. Постройте и оцените модель классификации с помощью перцептрона на датасете, который вы использовали на контрольной по классификиации (если вы ее не выполняли, возьмите любой датасет из раздела "real world datasets" в библиотеке sklearn).

# 5
Убедитесь, что алгоритм градиентного спуска завершается всегда - либо если ошибка перестает существенно меняться, либо при достижении максимального количества шагов.

```python
class Model(object):
    """Модель парной линейной регрессии"""
    def __init__(self):
        self.b0 = 0
        self.b1 = 0
    def predict(self, X):
        return self.b0 + self.b1 * X
    def error(self, X, Y):
        return sum(((self.predict(X) - Y)**2) / (2 * len(X)))
    def fit(self, X, Y=None, alpha=1.0, accuracy=0.01, max_steps=5000):
        if not(type(Y) == pd.Series):
            Y = X.iloc[:,1]
            X = X.iloc[:,0]
        steps, errors = [], []
        step = 0
        err = hyp.error(X, Y)
        for _ in range(max_steps):
            dJ0 = sum(self.predict(X) - Y) /len(X)
            dJ1 = sum((self.predict(X) - Y) * X) /len(X)
            self.b0 -= alpha * dJ0
            self.b1 -= alpha * dJ1
            new_err = hyp.error(X, Y)

            if (new_err>err):
                print('The error is going up!')
                break
            else:
                alpha /= 2
            err = new_err

            step += 1
            steps.append(step)
            errors.append(err)
        return steps, errors

    def draw_regress(self,x,y=None,alpha=1.0, accuracy=0.01, max_steps=5000):
        if not(type(y) == pd.Series):
            y = x.iloc[:,1]
            x = x.iloc[:,0]
        steps, errors = hyp.fit(x,y,alpha, accuracy, max_steps)
        J = hyp.error(x, y)
        print("error after gradient descent:", J)
        X0 = np.linspace(round(x.min())-1, round(x.max())+1, round(x.max()-x.min())*100)
        Y0 = hyp.predict(X0)
        plt.figure()
        plt.scatter(x, y)
        plt.plot(X0, Y0, 'r')
        plt.show()
```

```python
hyp = Model()
hyp.draw_regress(x,y)
```

```python
hyp = Model()
hyp.draw_regress(x,y,alpha=6)
```
